Let $A=\{x \in R | -2 \leq x \leq 8\}$ and $B=\{x \in R |3 < x < 11 \}$ 

Find:
 
a) $ A \cap B$

b) $ A \cup B$



Solution:

a) $ A \cap B = \{x \in R | -2 \leq x \leq 8$ and $3 < x < 11\} = \{x \in R |3 < x < 8 \}$

b) $ A \cup B = \{x \in R | -2 \leq x \leq 8$ or $3 < x < 11\} = \{x \in R |-2 < x < 11 \}$
